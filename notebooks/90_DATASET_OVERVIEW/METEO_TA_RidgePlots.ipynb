# **Imports**

```python
import importlib.metadata
import warnings
from datetime import datetime
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from diive.core.io.files import save_parquet, load_parquet
from diive.core.plotting.cumulative import CumulativeYear
from diive.core.plotting.heatmap_datetime import HeatmapDateTime, HeatmapYearMonth
from diive.core.plotting.xridgeplot import RidgePlotTS

warnings.filterwarnings(action='ignore', category=FutureWarning)
warnings.filterwarnings(action='ignore', category=UserWarning)
version_diive = importlib.metadata.version("diive")
print(f"diive version: v{version_diive}")
```

# **Load data**

```python
SOURCEDIR = r"../60_MERGE_DATA_FLUXES"
FILENAME = r"61.1_FLUXES_M10_MGMT_L4.1_NEE_LE_H_FN2O_FCH4.parquet"
FILEPATH = Path(SOURCEDIR) / FILENAME
df = load_parquet(filepath=FILEPATH)
df
```

# **Air temperature**

```python
series = df['TA_T1_2_1'].copy()
series
```

## All years (2005-2024)

### Per year (2005-2024)

```python
rp = RidgePlotTS(series=series)
rp.per_year(xlim=[-15, 35], ylim=[0, 0.14], hspace=-0.8, xlabel="Air temperature (°C)",
            fig_width=5, fig_height=9, shade_percentile=0.5, show_mean_line=False,
            fig_title="Air temperature per year (2005-2024)", fig_dpi=72)
```

### Per month (2005-2024)

```python
rp.per_month(xlim=[-15, 35], ylim=[0, 0.14], hspace=-0.6, xlabel="Air temperature (°C)",
             fig_width=4.5, fig_height=8, shade_percentile=0.5, show_mean_line=True,
             fig_title="Air temperatures per month (2005-2024)", fig_dpi=72)
```

### Per week (2005-2024)

```python
rp.per_week(xlim=[-15, 35], ylim=[0, 0.15], hspace=-0.6, xlabel="Air temperature (°C)",
            fig_width=6, fig_height=16, shade_percentile=0.5, show_mean_line=False,
            fig_title="Air temperatures per week (2005-2024)", fig_dpi=72)
```

# Single years per month

```python
uniq_years = series.index.year.unique()
for uy in uniq_years:
    series_yr = series.loc[series.index.year == uy].copy()
    rp = RidgePlotTS(series=series_yr)
    rp.per_month(xlim=[-15, 35], ylim=[0, 0.18], hspace=-0.6, xlabel="Air temperature (°C)",
                 fig_width=6, fig_height=7, shade_percentile=0.5, show_mean_line=False,
                 fig_title=f"Air temperatures per month ({uy})", fig_dpi=72)
```

# Single years per week

```python
uniq_years = series.index.year.unique()
for uy in uniq_years:
    series_yr = series.loc[series.index.year == uy].copy()
    rp = RidgePlotTS(series=series_yr)
    rp.per_week(xlim=[-15, 38], ylim=[0, 0.3], hspace=-0.8, xlabel="Air temperature (°C)",
                 fig_width=9, fig_height=18, shade_percentile=0.5, show_mean_line=False,
                 fig_title=f"Air temperatures per week ({uy})", fig_dpi=72)
```
