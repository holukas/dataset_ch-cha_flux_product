```python
import os
from pathlib import Path

def generate_file_links(folder_path, output_file):
    """Generate Markdown links to all files in a folder and writes them to a file."""

    with open(output_file, 'w') as f:
        f.write("# Notebook overview\n")
        f.write(f"## {folder_path}\n")
        for filename in sorted(os.listdir(folder_path)): #sorted for consistent order
            file_path = os.path.join(folder_path, filename)
            if os.path.isfile(file_path):
                # Create a relative path for the link
                relative_path = os.path.relpath(file_path, os.path.dirname(output_file))

                link = Path()
                
                link = f"- [{filename}]({relative_path})\n"
                f.write(link)

                
                print(relative_path)
                print(Path(file_path).stem)
                

# Example usage:
folder_to_scan = "00_L0_checks"
output_markdown = "README.md"
generate_file_links(folder_to_scan, output_markdown)

print(f"\nLinks generated and saved to {output_markdown}")
```
