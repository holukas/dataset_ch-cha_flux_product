```python
import os
import yaml

def generate_toc(root_dir=".", toc_file="_toc.yml"):
    """Generates _toc.yml from .ipynb files in the directory."""

    notebooks = sorted([f for f in os.listdir(root_dir) if f.endswith(".ipynb")])
    toc = []

    for nb in notebooks:
        title = os.path.splitext(nb)[0].replace("_", " ").title()  # Clean up title
        toc.append({"file": nb})

    with open(toc_file, "w") as f:
        yaml.dump(toc, f, default_flow_style=False)

    print(f"Generated _toc.yml with {len(notebooks)} entries.")

if __name__ == "__main__":
    generate_toc()
```
