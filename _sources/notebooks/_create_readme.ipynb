# Generate link to HTML notebooks for Jupyter book

```python
import os
from pathlib import Path
from diive.core.io.filereader import search_files

LOCATION = "https://holukas.github.io/dataset_ch-cha_flux_product/notebooks/"
           "https://holukas.github.io/dataset_ch-cha_flux_product/blob/gh-pages/notebooks/40_FLUX_PROCESSING_CHAIN/41.0_FluxProcessingChain_L3.3_NEE_QCF10.html"

notebooks = search_files(searchdirs="./", pattern="*.ipynb")
notebooks = [n for n in notebooks if ".ipynb_checkpoints" not in str(n)]
notebooks = [n for n in notebooks if "x-" not in str(n)]
notebooks = [n for n in notebooks if "tests" not in str(n)]
notebooks = [n for n in notebooks if not str(n.stem).startswith("_")]

chapter = ""

with open("README.md", 'w') as f:
    f.write("# Notebook overview\n")
    
    for nb in notebooks:   
        
        rel_filepath = nb.parent
        this_chapter = rel_filepath        
        
        if this_chapter != chapter:
            chapter = this_chapter
            f.write(f"\n\n## {chapter}\n")
            
        file_extension = nb.suffix      
        filename = nb.name
        filename_noext = nb.stem               
        filelink = Path(LOCATION) / rel_filepath / f"{filename_noext}.html"
        filelink = filelink.as_posix()  # Browser for html uses forward slash, also on Windows
        link = f"- [{filename}]({filelink})\n"
        f.write(link)               
        print(link)                                            

print(f"\nLinks generated and saved to {output_markdown}")
```
